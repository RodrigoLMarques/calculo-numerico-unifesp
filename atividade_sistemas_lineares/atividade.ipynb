#Parte 3

```python
import numpy as np


def eliminacao_gauss(A, b):
    """Resolve Ax = b por eliminação de Gauss SEM pivoteamento."""
    A = np.array(A, dtype=float)
    b = np.array(b, dtype=float)
    n = len(b)

    Ab = np.hstack((A, b.reshape(-1, 1)))

    for k in range(n - 1):
        pivo = Ab[k, k]
        if abs(pivo) < 1e-12:
            raise ZeroDivisionError(
                f"Pivô nulo na posição ({k},{k}). "
                "Sem pivoteamento, o método não pode continuar."
            )
        for i in range(k + 1, n):
            m = Ab[i, k] / pivo          # multiplicador
            Ab[i, k:] -= m * Ab[k, k:]   # Li <- Li - m * Lk

    if abs(Ab[n - 1, n - 1]) < 1e-12:
        raise ZeroDivisionError("Pivô nulo na última linha: sistema singular.")

    x = np.zeros(n)
    for i in range(n - 1, -1, -1):
        soma = np.dot(Ab[i, i + 1:n], x[i + 1:n])
        x[i] = (Ab[i, n] - soma) / Ab[i, i]

    return x, Ab

A = [
    [2, 1, -1],
    [-3, -1, 2],
    [-2, 1, 2]
    ]
b = [8, -11, -3]

x, Ab = eliminacao_gauss(A, b)

# np.set_printoptions(precision=6, suppress=True)
print("Matriz aumentada após a eliminação:")
print(Ab)
print("\nEliminação de Gauss implementada:", x)
print("Solução com np.linalg.solve:     ", np.linalg.solve(A, b))
```
